```python
import pandas as pd
import scipy.stats as stats

df=pd.read_csv('Employee.csv')
print(df.columns.tolist())
```

```python
stat, p_value = stats.shapiro(df['Performance_Score'].dropna())
print(f"P-value de Shapiro-Wilk : {p_value}")
if p_value > 0.05:
    print("Les données suivent une distribution normale.")
else:
    print("Les données ne suivent pas une distribution normale.")
```

```python
tableau_contingence = pd.crosstab(df['Department'], df['Employee_Satisfaction_Score'])
chi2, pval, dof, exepected = stats.chi2_contingency(tableau_contingence)
print(f"P-value du khi2 : {pval}")
if pval < 0.05:
    print("Il y a une relation significative entre les variables 'Department' et 'Employee_Satisfaction_Score'")
else :
    print("La satisfaction est indépendante du département (Pas de lien prouvé).")
```

```python
from scipy.stats.contingency import association
v_cramer = association(tableau_contingence, method='cramer')
print(f"Score V de cramer : {v_cramer:.3}")
if v_cramer < 0.1:
  print ("lien négligeable")
elif v_cramer < 0.3:
    print ("lien faible")
elif v_cramer < 0.5:
      print ("lien moyen")
else:
        print ("lien fort")
```

```python
import scipy.stats as stats
scores_par_dept = [df[df['Department'] == dept]['Performance_Score'].dropna() for dept in df['Department'].unique()]
stat_levene, p_levene = stats.levene(*scores_par_dept)
print(f"P-value de Levene : {p_levene:.4f}")
if p_levene > 0.05:
    print("Les variances sont similaires.")
else:
    print("Les variances sont différentes")
print("-"*50)
stat_kruskal, p_kruskal = stats.kruskal(*scores_par_dept)
print(f"P-value de Kruskal-Wallis : {p_kruskal:.4f}")
if p_kruskal > 0.05:
    print("tous les départements ont un niveau de performance similaire")
else:
    print("Il y a une vraie différence de performance entre les départements")
```

```python
colonne_quanti_1 = 'Work_Hours_Per_Week'
colonne_quanti_2 = 'Performance_Score'
df_clean = df[[colonne_quanti_1, colonne_quanti_2]].dropna()

stat_spearman, p_spearman = stats.spearmanr(df_clean[colonne_quanti_1], df_clean[colonne_quanti_2])

print(f"Coefficient de corrélation de Spearman : {stat_spearman:.4f}")
print(f"P-value de Spearman : {p_spearman:.4f}")

print("_"*50)
if p_spearman > 0.05:
    print("Il n'y a pas de relation linéaire entre ces deux variables.")
else:
    print("Il y a une relation linéaire entre ces deux variables.")
```

L'analyse statistique des données RH met en évidence plusieurs dynamiques essentielles pour l'entreprise. Les tests confirment tout d'abord que la satisfaction des collaborateurs est directement liée à leur département, soulignant l'impact probable du management local ou des conditions spécifiques à chaque service sur le climat de travail. Cette disparité se retrouve également dans les évaluations individuelles : le test de Kruskal-Wallis prouve qu'il existe un véritable écart de notation entre les différents départements, ce qui indique un manque d'harmonisation ou une sévérité inégale dans les critères d'évaluation des managers. Enfin, l'étude des corrélations démontre que le volume d'heures travaillées par semaine n'a aucune incidence statistique sur les performances réelles des employés. Fort de ces constats, il est recommandé de mettre en place des sessions de calibrage managérial pour garantir l'équité des futures évaluations, de cibler les départements à faible satisfaction pour comprendre leurs points de blocage, et d'orienter définitivement la culture d'entreprise vers l'atteinte d'objectifs qualitatifs plutôt que sur le simple présentéisme.
